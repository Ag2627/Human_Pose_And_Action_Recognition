```python
import os
import subprocess
import sys
import time
import socket

PROJECT_DIR = r"C:\Users\admin\ViTPose_Action_Recognition"
FRONTEND_DIR = os.path.join(PROJECT_DIR, "frontend")

os.chdir(FRONTEND_DIR)

# Start Streamlit in the background
log_file = open("streamlit_log.txt", "w")

process = subprocess.Popen(
    [
        sys.executable,
        "-m",
        "streamlit",
        "run",
        "app.py",
        "--server.port",
        "8501",
        "--server.headless",
        "true"
    ],
    stdout=log_file,
    stderr=subprocess.STDOUT,
    cwd=FRONTEND_DIR
)

print("Streamlit process started.")
print("Process ID:", process.pid)
```

```python
import socket
import time

time.sleep(5)

sock = socket.socket()
result = sock.connect_ex(("127.0.0.1", 8501))
sock.close()

if result == 0:
    print("SUCCESS: Streamlit is running!")
    print("Open this address: http://localhost:8501")
else:
    print("Streamlit has not opened its port yet.")
```
